# Cálculo de Volumes

Neste notebook, calcularemos:
1. Volume da Esfera: V = (4/3) × π × r³
2. Volume do Cilindro: V = π × r² × h

```python
import math

# Volume da Esfera
def volume_esfera(raio):
    """
    Calcula o volume de uma esfera.
    Fórmula: V = (4/3) × π × r³
    """
    return (4/3) * math.pi * (raio ** 3)

# Volume do Cilindro (raio × comprimento/altura)
def volume_cilindro(raio, altura):
    """
    Calcula o volume de um cilindro.
    Fórmula: V = π × r² × h
    """
    return math.pi * (raio ** 2) * altura

# Exemplos de uso
raio_esfera = 5  # em cm
raio_cilindro = 3  # em cm
altura_cilindro = 10  # em cm

v_esfera = volume_esfera(raio_esfera)
v_cilindro = volume_cilindro(raio_cilindro, altura_cilindro)

print(f"Volume da Esfera (r = {raio_esfera} cm): {v_esfera:.2f} cm³")
print(f"Volume do Cilindro (r = {raio_cilindro} cm, h = {altura_cilindro} cm): {v_cilindro:.2f} cm³")
```

```python
# Teste interativo com diferentes valores
print("\n=== Teste de Volumes ===")

# Esfera com raio 2
print(f"\nEsfera com raio 2 cm: {volume_esfera(2):.2f} cm³")

# Esfera com raio 7
print(f"Esfera com raio 7 cm: {volume_esfera(7):.2f} cm³")

# Cilindro com raio 2 e altura 5
print(f"\nCilindro com raio 2 cm e altura 5 cm: {volume_cilindro(2, 5):.2f} cm³")

# Cilindro com raio 4 e altura 12
print(f"Cilindro com raio 4 cm e altura 12 cm: {volume_cilindro(4, 12):.2f} cm³")
```
